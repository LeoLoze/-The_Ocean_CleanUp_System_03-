**Removed:** `animal_density` (2.1e-3 per m³, derived as 1/180 of the plastic pieces). This density describes **neuston** (small floating organisms such as *Velella* and copepods) at the sea surface, but it was applied to the whole swept water volume with a 63 % capture probability. This gave ~9e7 animals per year for one system, about 4,000 times more than observed. Neuston is not bycatch: the EIA and Egger et al. (2025) found no systematic impact on neuston.

### Calibration of `bycatch_ref`

The absolute bycatch level is calibrated on the monitoring data of System 002, campaigns 1–12 (July 2021 – December 2022):

| Quantity | Value | Source |
|---|---|---|
| Fish and sharks caught (primary bycatch) | 13,789 individuals (mainly blennies 6,272, sergeant majors 3,652, pygmy sharks 2,122) | CSA Ocean Sciences Inc. (2023), Table 5-7 |
| Sea turtles captured (10 dead, 6 released alive) | 16 individuals | CSA Ocean Sciences Inc. (2023), §5.2.5; The Ocean Cleanup (2023b) |
| **Total primary bycatch** | **~13,800 individuals** | |
| Plastic removed in the same period | 209.6 t (primary bycatch = 0.33 % of the plastic weight) | CSA Ocean Sciences Inc. (2023), §5.2.3.2 |
| Nominal span of System 002 | 520 m | CSA Ocean Sciences Inc. (2023), §2.1.1.1 |
| Operating time | 1.5 yr × `uptime` (0.7) = 3.3e7 s | Assumption: the EIA gives campaign lengths (~6 weeks incl. 4–6 days transit each way), not towing hours |
| Area swept | 520 m × 0.75 m/s × 3.3e7 s ≈ 12,900 km² | |

$$b_{ref} = \frac{13{,}800}{12{,}900\ \text{km}^2} \approx 1.0\ \text{animal per km}^2\ \text{swept}$$

**Plausibility checks**
- The EIA addendum (Table 7, densities from Egger et al., 2021) reports 622–4,949 fish per km² in the garbage patch. Only ~1 per km² is caught, i.e. ~0.02–0.2 % of the fish in the swept path. This matches the EIA observation that "numerous fish were observed on cameras swimming into and out of the System through the mesh".
- Today's design (`X_irl`: one System 03, 4 m skirt, 1600 m span, 0.70 m/s, 16 mm mesh) gives ~20,000 animals/yr. Scaling the observed 66 animals per tonne of plastic (13,800 / 209.6 t) with the modelled removal of ~456 t/yr gives ~30,000 animals/yr, which is the same order of magnitude.

**Limitations**
- Only the absolute level is calibrated; the depth, speed and mesh factors are relative shapes (all equal to 1 for System 002) and are not calibrated individually.
- All individuals count equally (a blenny counts as much as a turtle). Turtles are ~0.1 % of the count.
- The calibration uses System 002; System 03 has a different retention zone and a coarser mesh (partly captured by the mesh factor).
