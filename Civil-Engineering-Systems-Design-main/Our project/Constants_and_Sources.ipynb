## Calibration of the drag coefficient (`drag_coeff`, `Cd`)

$C_D$ is an effective coefficient for the whole towed system in $F = \tfrac{1}{2}\rho_w C_D (x_1 x_2) x_4^2$, calibrated on the towline tensions measured on System 002 between August 2021 and August 2022:

| Quantity | Value | Source |
|---|---|---|
| Measured quantity | Combined tension of both towing lines, 3-hour means | Gonzalez Jimenez et al. (2023), Section 2.1 |
| System 002 | 800 m long, 3 m draft, 10 mm mesh | Gonzalez Jimenez et al. (2023), Section 1 and Table 1 |
| Span 480 m | 10 points, 0.51–0.86 m/s, 13–38 tf | Gonzalez Jimenez et al. (2023), Figure 19 |
| Span 630 m | 8 points, 0.50–0.83 m/s, 17–53 tf | Gonzalez Jimenez et al. (2023), Figure 19 |

Each point gives $C_D = T / (\tfrac{1}{2}\rho_w \cdot 3\,\text{m} \cdot \text{span} \cdot u^2)$. A least-squares fit through all 18 points gives $C_D \approx 0.72$ (0.67 for 480 m, 0.75 for 630 m), so $C_D = 0.7$.

- The paper's OrcaFlex model gives 0.65–0.67 in still water and about 0.75 with its wave factor of 1.15.
- Today's design gives ~590 kN (60 tf) per towing line, about a third of the bollard pull.
- The points are read from the figure and cover 0.50–0.86 m/s (bounds up to 1.29 m/s). The 10 mm mesh of System 002 makes the value conservative for the 16 mm mesh of System 03.

## Depth factor of the bycatch

The bycatch scales with $\sqrt{x_1/d_{ref}}$ instead of linearly with the skirt depth, because `bycatch_ref` is per km² of sea surface swept and deeper wings increase the bycatch only "to a slightly greater extent than S002" (CSA Ocean Sciences Inc., 2023).
