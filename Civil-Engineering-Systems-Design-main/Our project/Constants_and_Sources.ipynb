### Depth factor of the bycatch

The bycatch scales with $\sqrt{x_1/d_{ref}}$, not linearly with the skirt depth:

- The bycatch is calibrated per km² of sea *surface* swept, because it consists mainly of fish that live near the surface around floating debris. A linear depth factor would turn the swept area back into a swept volume.
- For the step from 3 m to 4 m wings (System 03), the EIA expects fish to be caught only "to a slightly greater extent than S002 due to the deeper wings, which could decrease the ability for some fish to escape the wings. It could be more difficult for some fish species to swim under the deeper wings to escape, but most fish species that have been included in the S002 bycatch would not be affected" (CSA Ocean Sciences Inc., 2023).

### Calibration of the drag coefficient (`drag_coeff`, `Cd`)

The cost objective and constraint 1 use the same drag formula, $F = \tfrac{1}{2}\rho_w C_D (x_1 x_2) x_4^2$, with skirt depth &times; span as the reference area. $C_D$ is therefore an *effective* coefficient for the whole towed system (net, floats, ballast line and retention zone in a U-shape), not the drag coefficient of a single twine or of a flat net panel. It is calibrated on the towline tensions that The Ocean Cleanup measured on System 002 in the Great Pacific Garbage Patch between August 2021 and August 2022:

| Quantity | Value | Source |
|---|---|---|
| Measured quantity | Combined towline tension (sum of both towing lines), 3-hour means, measured with a Running Line Monitor on each towing line | Gonzalez Jimenez et al. (2023), §2.1 |
| System 002 | 800 m long, netting draft 3 m, 10 mm mesh | Gonzalez Jimenez et al. (2023), §1 and Table 1 |
| Nominal span, 480 m | 10 data points, 0.51&ndash;0.86 m/s, 13&ndash;38 tf (e.g. ~30 tf at 0.77 m/s) | Gonzalez Jimenez et al. (2023), Fig. 19 |
| Wide span, 630 m | 8 data points, 0.50&ndash;0.83 m/s, 17&ndash;53 tf (e.g. ~38 tf at 0.75 m/s) | Gonzalez Jimenez et al. (2023), Fig. 19 |

For every data point, $C_D = T / (\tfrac{1}{2}\rho_w \cdot 3\,\text{m} \cdot \text{span} \cdot u^2)$, with $T$ the combined towline tension and $u$ the speed through water. A least-squares fit through all 18 points gives

$$C_D \approx 0.72 \quad (0.67 \text{ for the 480 m span},\ 0.75 \text{ for the 630 m span}) \quad\Rightarrow\quad C_D = 0.7$$

**Plausibility checks**
- The single points scatter between 0.58 and 1.07, because the sea state differs between the measurements. The validated OrcaFlex model of the paper, which is in still water, gives 0.65&ndash;0.67 at 0.75 m/s. The paper applies a wave factor of 1.15 for median sea states, which gives about 0.75.
- The previous value of 1.2 is in the range of the drag coefficient of a single twine (a cylinder). Applied to depth &times; span, it overestimates the measured towline tension by about 70 %.
- Today's design (`X_irl`: 4 m skirt, 1460 m span, 0.75 m/s) gives a total drag of ~1,180 kN, i.e. ~590 kN (60 tf) per towing line. This is about a third of the 173 t bollard pull of one towing vessel.

**Limitations**
- The data points were read from the published figure (a vector graphic), not from a data table.
- The calibration uses System 002 with a 10 mm mesh. System 03 has a more open 16 mm mesh, so the value is probably on the high (safe) side for today's design. The effect of the mesh size (`x5`) on the drag is not modelled.
- Depth &times; span as reference area assumes a span-to-length ratio like in the measurements (0.6 and 0.8). The span bounds of 520&ndash;1800 m for the 2200 m barrier give ratios of 0.24&ndash;0.82. The measurements cover 0.50&ndash;0.86 m/s; the speed bounds go up to 1.29 m/s.
